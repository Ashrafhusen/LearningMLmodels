```python
from transformers import BertTokenizer, BertModel
import torch
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
text = "The river bank was full of reeds."
input = tokenizer(text, return_tensors= 'pt')
outputs = model(**input)
```

```python
token_ids = tokenizer.convert_tokens_to_ids(input)
```

```python
print("Token IDs:", token_ids)
```

```python
print("Tokens:", input)
```

```python
sentence_embedding = outputs.last_hidden_state[0][0]
print(sentence_embedding)
```

```python
from sentence_transformers import SentenceTransformer, util

model_sbert = SentenceTransformer('all-MiniLM-L6-v2')

sentences = [
    "The Eiffel Tower is in Paris.",
    "I visited France last summer.",
    "Dogs are great pets."
]

```

```python
embeddings = model_sbert.encode(sentences, convert_to_tensor=True)
similarity = util.cos_sim(embeddings[0], embeddings)
print(f"is : {similarity}")
```
